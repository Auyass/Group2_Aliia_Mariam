## Part A: ReviewClassifier class
The class stores positive and negative words. The method classify() counts them and decides if a review is positive, negative or neutral.

```python
class ReviewClassifier:
    def __init__(self):
        self.positive_words = ["good", "great", "love", "excellent", "nice"]
        self.negative_words = ["bad", "terrible", "boring", "hate", "awful"]
    def classify(self, review):
        words = review.lower().split()
        score = 0
        for word in words:
            if word in self.positive_words:
                score += 1
            elif word in self.negative_words:
                score -= 1
        # decide after the loop is finished
        if score > 0:
            return "positive"
        elif score < 0:
            return "negative"
        else:
            return "neutral"
```

## Part B: Quick check
We test the classifier on two simple reviews.

```python
classifier = ReviewClassifier()
print(classifier.classify("This movie was great and I love it"))
print(classifier.classify("Terrible and boring"))
```

```python
classifier = ReviewClassifier()
reviews = ["great food and nice staff", "I love this excellent movie", "boring and bad film", "terrible service I hate it", "the film was two hours long", "not good at all"]
correct_answers = ["positive", "positive", "negative", "negative", "neutral", "negative"]
correct = 0
for i in range(len(reviews)):
    result = classifier.classify(reviews[i])
    print(reviews[i], "->", result)
    if result == correct_answers[i]:
        correct += 1
print(f"Correct: {correct} out of {len(reviews)}")
```

## Part C: Testing accuracy
We test on 6 reviews where we know the correct answer and count how many are right.
