```python
import random
suits = ["Hearts", "Diamonds", "Clubs", "Spades"]
values = ["Ace", 2, 3, 4, 5, 6, 7, 8, 9, 10, "Jack", "Queen", "King"]
deck = []
for suit in suits:
  for value in values:
    deck.append((value, suit))
random.shuffle(deck)
hand=deck[:5]
hcount=0
for value, suit in hand:
  print(value, "of", suit)
  if suit=="Hearts":
    hcount +=1
print("Heart count=",hcount)
```
