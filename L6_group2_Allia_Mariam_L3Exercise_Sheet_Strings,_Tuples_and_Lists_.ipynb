Stretch 2. Average mark

```python
marks = [80, 90, 75, 88, 92, 85]

total = 0

for mark in marks:
    total += mark

average = total / len(marks)

highest = marks[0]

for mark in marks:
    if mark > highest:
        highest = mark

print("Average:", average)
print("Highest mark:", highest)
```

Stretch 3. Palindrome check

```python
word = input("Enter a word: ")

if word == word[::-1]:
    print("Palindrome!")
else:
    print("Not a palindrome.")
```

Stretch 4. Limited guesses

```python
import random

secret = random.randint(1, 100)
guesses = 0

while True:
    guess = int(input("Guess a number (1-100): "))
    guesses += 1

    if guess == secret:
        print("Correct!")
        print("You needed", guesses, "guesses.")
        break

    elif guess < secret:
        print("Higher")

    else:
        print("Lower")

    if guesses == 7:
        print("You lost! The number was", secret)
        break
```
